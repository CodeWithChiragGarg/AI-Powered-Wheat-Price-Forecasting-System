```python
df['Arrival Quantity'] = df['Arrival Quantity'].astype(str).str.replace(',', '').astype(float)

X_multi = df[['Day', 'Arrival Quantity']]
y = df['Modal Price']

from sklearn.linear_model import LinearRegression

model_multi = LinearRegression()
model_multi.fit(X_multi, y)

pred_multi = model_multi.predict(X_multi)

print("Multiple Linear Regression done")
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=150, random_state=42)
rf_model.fit(X_multi, y)

rf_pred = rf_model.predict(X_multi)

print("Random Forest model trained")
```

```python
from sklearn.metrics import mean_absolute_error

mae_lr = mean_absolute_error(y, pred_multi)
mae_rf = mean_absolute_error(y, rf_pred)

print("Linear Regression MAE:", mae_lr)
print("Random Forest MAE:", mae_rf)
```

```python
def predict_price_advanced(input_date):
    import pandas as pd

    input_date = pd.to_datetime(input_date)
    day_number = (input_date - df['Arrival Date'].min()).days

    avg_arrival = df['Arrival Quantity'].mean()

    input_df = pd.DataFrame({
        'Day': [day_number],
        'Arrival Quantity': [avg_arrival]
    })

    return rf_model.predict(input_df)[0]
```

```python
predict_price_advanced("2026-04-26")
```

```python
def predict_rate(date_input):
    import pandas as pd

    date = pd.to_datetime(date_input)
    day = (date - df['Arrival Date'].min()).days

    avg_arrival = df['Arrival Quantity'].mean()

    input_df = pd.DataFrame({
        'Day': [day],
        'Arrival Quantity': [avg_arrival]
    })

    price = rf_model.predict(input_df)[0]

    print(f"Predicted Wheat Price on {date_input} is: ₹{price:.2f} per Quintal")
```
